# PIVOT TABLE

## It's like in Excel, but with Pandas

There are 2 syntaxes, choose your preferred :
- As a method : YOUR_DATAFRAME.pivot_table()
- As a function : pd.crosstab()

```python
# Example
df_weather.pivot_table(values = 'MAX_TEMPERATURE_C', index = 'MONTH', aggfunc = 'mean')
```

## Create a Pivot table with :
- Months on rows
- the maximum of Max Temperature on values
- Opinion on columns

```python
# Hop !
pivot_1 = df_weather.pivot_table(values='MAX_TEMPERATURE_C', index='MONTH', columns='OPINION', aggfunc='max')
pivot_1
```

## Pivot table with :
- Days on rows
- Precipitation on values
- Months on columns


```python
# Hop !
pivot_2 = df_weather.pivot_table(values='PRECIP_TOTAL_DAY_MM', index='DAY', columns='MONTH')
pivot_2
```

## Pivot table with :
- Months on rows
- the sum of Precipitations on values

**Then display it as a column chart.** (optionnal)

```python
# Hop !
pivot_3 = df_weather.pivot_table(values='PRECIP_TOTAL_DAY_MM', index='MONTH', aggfunc='sum')
pivot_3.plot(kind='bar')
```

## Pivot table with :
- Months on rows
- the average of Max Temperature on a first column of values
- the average of Sunhours on a second column of values

```python
# Hop !
pivot_4 = df_weather.pivot_table(values=['MAX_TEMPERATURE_C', 'SUNHOUR'], index='MONTH', aggfunc='mean')
pivot_4
```

# Chart
Using your favorite library (MatPlotLib, Seaborn, Plotly or what you want), draw a chart from the previous pivot table with :
- Temperature in a blue column chart
- Sunhours in a red line plot

Tip : Once your pivot table created, you can help yourself with the `.reset_index()`  method.

```python
# Your code here
import matplotlib.pyplot as plt

pivot_chart = pivot_4.reset_index()

fig, ax1 = plt.subplots(figsize=(10, 5))
ax1.bar(pivot_chart['MONTH'], pivot_chart['MAX_TEMPERATURE_C'], color='blue')
ax1.set_xlabel('Month')
ax1.set_ylabel('Max temperature (°C)')
ax1.set_xticks(range(1, 13))

ax2 = ax1.twinx()
ax2.plot(pivot_chart['MONTH'], pivot_chart['SUNHOUR'], color='red')
ax2.set_ylabel('Sunhours')

plt.show()
```
