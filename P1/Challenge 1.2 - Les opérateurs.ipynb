# 1. Calcul d'un prix

Ecrivez un programme faisant saisir à l’utilisateur un prix unitaire HT, un taux de TVA et un nombre d’articles, et qui calcule le montant TTC de l’achat.<br>
Affichez le montant TTC.

```python
def get_price_including_taxes(price: float, vta: float, n: int) -> float:
    return n * (price + (price * (vta / 100)))

original_price : float = float(input("Price : "))
vta : float = float(input("VTA : "))
article_n: int = int(input("Number of articles : "))
price = get_price_including_taxes(original_price, vta, article_n)

print(f"Price with VTA : {price}")
```

# 2. Conversion de durées (1)

Ecrivez un programme demandant une durée en secondes à l’utilisateur, et qui la convertit en heures, minutes, secondes.<br>
Affichez le résultat.<br>
Par exemple 12334 deviendra 3 heures, 25 minutes et 34 secondes.

```python
def convert_seconds(sec: int) -> tuple[int, int, int]:
    hours, remainder = divmod(sec, 3600)
    minutes, seconds = divmod(remainder, 60)
    return hours, minutes, seconds

sec: int = int(input("Secs : "))
h, m, s = convert_seconds(sec)

print(f"{sec} secondes = {h} heure(s), {m} minute(s) et {s} seconde(s)")
```

# 3. Conversion de durées (2)

Ecrivez un programme demandant une durée en heures, minutes, secondes à l’utilisateur et qui la convertit en secondes.<br>
Affichez le résultat.<br>
Par exemple 3 heures, 25 minutes et 34 secondes deviendra 12334 secondes.

```python
def convert_to_seconds(hours: int, minutes: int, seconds: int) -> int:
    return hours * 3600 + minutes * 60 + seconds

h: int = int(input("Heures : "))
m: int = int(input("Minutes : "))
s: int = int(input("Secondes : "))

total: int = convert_to_seconds(h, m, s)

print(f"{h} heure(s), {m} minute(s) et {s} seconde(s) = {total} secondes")
```
