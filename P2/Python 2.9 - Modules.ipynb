- Fonctions : `carre(x)`, `cube(x)`, `factorielle(n)`, `valeur_absolue(x)`, `maximum(liste)`, `minimum(liste)`, `moyenne(liste)`, `ecart_type(liste)`, `mediane(liste)`

```python
%%writefile data_utils.py
PI = 3.141592653589793
E = 2.718281828459045


def carre(x):
    """Return the square of x.

    Parameters: x (int or float)
    Returns: x * x
    """
    return x * x


def cube(x):
    """Return the cube of x.

    Parameters: x (int or float)
    Returns: x * x * x
    """
    return x * x * x


def factorielle(n):
    """Compute the factorial of a positive integer.

    Parameters: n (int), must be >= 0
    Returns: n! (int)
    """
    resultat = 1
    for i in range(2, n + 1):
        resultat *= i
    return resultat


def valeur_absolue(x):
    """Return the absolute value of x.

    Parameters: x (int or float)
    Returns: x if x >= 0, otherwise -x
    """
    if x < 0:
        return -x
    return x


def maximum(liste):
    """Return the biggest element of a list.

    Parameters: liste (list of numbers)
    Returns: the maximum value
    """
    max_val = liste[0]
    for elt in liste:
        if elt > max_val:
            max_val = elt
    return max_val


def minimum(liste):
    """Return the smallest element of a list.

    Parameters: liste (list of numbers)
    Returns: the minimum value
    """
    min_val = liste[0]
    for elt in liste:
        if elt < min_val:
            min_val = elt
    return min_val


def moyenne(liste):
    """Return the mean of a list of numbers.

    Parameters: liste (list of numbers)
    Returns: sum of the elements divided by their number (float)
    """
    return sum(liste) / len(liste)


def ecart_type(liste):
    """Return the (population) standard deviation of a list of numbers.

    Parameters: liste (list of numbers)
    Returns: the standard deviation (float)
    """
    m = moyenne(liste)
    variance = sum((x - m) ** 2 for x in liste) / len(liste)
    return variance ** 0.5


def mediane(liste):
    """Return the median of a list of numbers.

    Parameters: liste (list of numbers)
    Returns: the middle value (or the mean of the two middle values)
    """
    triee = sorted(liste)
    n = len(triee)
    milieu = n // 2
    if n % 2 == 1:
        return triee[milieu]
    return (triee[milieu - 1] + triee[milieu]) / 2
```

## Etape 1 : Refaites le process qu'on a vu ensemble dans les quêtes et au début du challenge pour créer une nouvelle version du module `data_utils_V2.py` en rajoutant de nouvelles fonctions.


*   Création du fichier `data_utils_V2.py`
*   Ajout et développement de nouvelles fonctions :
  * `somme_cumulee(liste)`: calcule la somme cumulée d'une liste et retourne une liste. Par exemple, pour la liste [1, 2, 3, 4], la somme cumulée serait [1, 3, 6, 10].

  * `compter_occurrences(liste)`: retourne un dictionnaire contenant le nombre d'occurrences de chaque élément de la liste. Par exemple, pour la liste [1, 2, 2, 3], le résultat serait {1: 1, 2: 2, 3: 1}.
  
  * `inverser_dict(dictionnaire)`: inverse les clés et les valeurs d'un dictionnaire. Par exemple, pour le dictionnaire {'a': 1, 'b': 2}, le résultat serait {1: 'a', 2: 'b'}.

  * `filtrer_pairs(liste)`: retourne une nouvelle liste contenant uniquement les nombres pairs de la liste d'entrée. Par exemple, pour la liste [1, 2, 3, 4], le résultat serait [2, 4].

  * `celsius_to_fahrenheit(celsius)`: convertit une température de Celsius en Fahrenheit et retourne un nombre. Par exemple, pour une température de 0°C, le résultat serait 32°F. *Rappel : la formule est F = (C * 9/5) + 32*

  * `trouver_doublons(liste)`: retourne une liste des éléments qui apparaissent plus d'une fois dans la liste d'entrée. Par exemple, pour la liste [1, 2, 2, 3, 3], le résultat serait [2, 3].

  * `calculer_moyenne_ponderee(valeurs, poids)`: calcule la moyenne pondérée d'une liste de valeurs avec une liste de poids correspondante et retourne un nombre. Par exemple, pour les valeurs [3, 5] et les poids [1, 2], le résultat serait 4.33.





```python
%%writefile data_utils_V2.py
from data_utils import *


def somme_cumulee(liste):
    """Compute the cumulative sum of a list.

    Parameters: liste (list of numbers)
    Returns: a list, e.g. [1, 2, 3, 4] -> [1, 3, 6, 10]
    """
    resultat = []
    total = 0
    for elt in liste:
        total += elt
        resultat.append(total)
    return resultat


def compter_occurrences(liste):
    """Count how many times each element appears in a list.

    Parameters: liste (list)
    Returns: a dictionary {element: number of occurrences}
    """
    occurrences = {}
    for elt in liste:
        occurrences[elt] = occurrences.get(elt, 0) + 1
    return occurrences


def inverser_dict(dictionnaire):
    """Swap the keys and the values of a dictionary.

    Parameters: dictionnaire (dict), the values must be hashable
    Returns: a new dictionary {value: key}
    """
    return {valeur: cle for cle, valeur in dictionnaire.items()}


def filtrer_pairs(liste):
    """Keep only the even numbers of a list.

    Parameters: liste (list of int)
    Returns: a new list with the even numbers
    """
    return [x for x in liste if x % 2 == 0]


def celsius_to_fahrenheit(celsius):
    """Convert a temperature from Celsius to Fahrenheit (F = C * 9/5 + 32).

    Parameters: celsius (int or float)
    Returns: the temperature in Fahrenheit (float)
    """
    return celsius * 9 / 5 + 32


def trouver_doublons(liste):
    """Find the elements that appear more than once in a list.

    Parameters: liste (list)
    Returns: a list of the duplicated elements (each one only once)
    """
    doublons = []
    for elt, nb in compter_occurrences(liste).items():
        if nb > 1:
            doublons.append(elt)
    return doublons


def calculer_moyenne_ponderee(valeurs, poids):
    """Compute the weighted mean of a list of values.

    Parameters: valeurs (list of numbers), poids (list of numbers, same length)
    Returns: sum(valeur * poids) / sum(poids) (float)
    """
    somme = 0
    for v, p in zip(valeurs, poids):
        somme += v * p
    return somme / sum(poids)
```

# Partie Bonus

Créez une fonction `analyser_texte(texte)` qui prend une chaîne de caractères en entrée et retourne un dictionnaire contenant les informations suivantes :

- Nombre total de mots
- Nombre de caractères (espaces inclus)
- Nombre de caractères (espaces exclus)
- Mot le plus long
- Longueur moyenne des mots

Testez cette fonction avec le texte suivant :
"Python est un langage de programmation puissant et facile à apprendre. Il dispose de structures de données de haut niveau efficaces et d'une approche simple mais efficace de la programmation orientée objet."

```python
def analyser_texte(texte):
    mots = texte.split()
    mots_propres = [mot.strip(".,;:!?") for mot in mots]
    mot_le_plus_long = max(mots_propres, key=len)
    return {
        "nombre_de_mots": len(mots),
        "caracteres_avec_espaces": len(texte),
        "caracteres_sans_espaces": len(texte.replace(" ", "")),
        "mot_le_plus_long": mot_le_plus_long,
        "longueur_moyenne_des_mots": round(sum(len(m) for m in mots_propres) / len(mots_propres), 2),
    }


texte = "Python est un langage de programmation puissant et facile à apprendre. Il dispose de structures de données de haut niveau efficaces et d'une approche simple mais efficace de la programmation orientée objet."
analyser_texte(texte)
```
